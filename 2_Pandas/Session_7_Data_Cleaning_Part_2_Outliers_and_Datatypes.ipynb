1. Given a CSV file containing Zomato restaurant ratings, use pandas to detect outliers in the 'user_rating' column using the IQR method and print the indices of the detected outliers.


```python
dataset = pd.read_csv("F:\Tops_assignment\Python_Library\datasets\zomato.csv\zomato.csv")
```

```python
dataset.head()
```

```python
dataset.info()
```

```python

dataset["rate"] = dataset["rate"].str.replace("/5" ,"")
dataset.head()
dataset["rate"] = pd.to_numeric(dataset["rate"] , errors= "coerce")
dataset["rate"].info()
```

```python
dataset["rate"].info()
```

```python
#IQR method 

q1_value = dataset["rate"].quantile(0.25)
q2_value = dataset["rate"].quantile(0.50)
q3_value = dataset["rate"].quantile(0.75)

print( q1_value , q2_value, q3_value)

IQR = 1.5 * (q3_value - q1_value)

lowerb = q1_value - IQR
upperb = q3_value + IQR

print("\n" , lowerb , upperb)

outliers =((dataset["rate"] < lowerb ) | (dataset["rate"] > upperb)).sum()


outliers1 =  dataset["rate"][(dataset["rate"] < lowerb ) | (dataset["rate"] > upperb)]

null_Values = dataset["rate"].isnull().sum()
print(f"we have total {outliers} outliers" )

print(f"we have total {null_Values} null values" )

print(outliers1.index)
```

3. Apply winsorization to the 'transaction_amount' column in a Paytm transactions DataFrame to cap all values above the 95th percentile and below the 5th percentile, then display the updated column statistics.


```python
dataset1 = pd.read_csv("F:\Tops_assignment\Python_Library\datasets\paytm_transactions_100_with_outliers.csv")
dataset1.info()
```

```python


q1_value = dataset1["Transaction_Amount"].quantile(0.05)
q2_value = dataset1["Transaction_Amount"].quantile(0.50)
q3_value = dataset1["Transaction_Amount"].quantile(0.95)

print( q1_value , q2_value, q3_value)

IQR = 1.5 * (q3_value - q1_value)

lowerb = q1_value - IQR
if lowerb < 0:
    lowerb = 0
upperb = q3_value + IQR


print("\n" , lowerb , upperb)



print(dataset1["Transaction_Amount"].describe())

outliers = dataset1["Transaction_Amount"][(dataset1["Transaction_Amount"]<lowerb) | (dataset1["Transaction_Amount"]>upperb)]


dataset1["Transaction_Amount"] = dataset1["Transaction_Amount"].clip(lower= lowerb , upper=upperb)

print(dataset1["Transaction_Amount"].describe())
```


5. Fix the following code snippet where the 'is_premium' column in a Spotify user DataFrame is a mix of boolean, string, and integer types. Convert the entire column to boolean type, treating 'True', 1, and 'yes' as True,
and everything else as False.

```python
import numpy as np
```

```python
np.random.seed(42)

df = pd.DataFrame({
    "user_id": range(1, 101),
    "is_premium": np.random.choice(
        [True, False, "yes", "no", 1, 0],
        size=100
    )
})

print(df.head(10))
```

```python
df["is_premium"] = df["is_premium"].str.replace("yes" ,"True")
df["is_premium"] = df["is_premium"].str.replace("True" ,"1")
print(df.head(10))

```

```python
## best way 

df["is_premium"] = df["is_premium"].apply(
    lambda x: str(x).lower() in ["true", "1", "yes"]
)

print(df.head(15))
```
