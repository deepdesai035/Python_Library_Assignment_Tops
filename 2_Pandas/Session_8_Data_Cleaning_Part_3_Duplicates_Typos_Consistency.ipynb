
4. Suppose you have a DataFrame of Instagram usernames where some entries have typos (like 'insta_queen', 'insta-queen', 'instaqueen'). Use the replace() function to standardize all these variants to 'instaqueen'.


```python
list1 = ["insta_queen" , "insta-queen ", "instaqueen"]*100

df = pd.DataFrame({"val":list1})

print(df)

df["val"] = df["val"].str.replace("_","")
df["val"] = df["val"].str.replace("-","")
df["val"] = df["val"].str.replace(" ","")
df["val"]
```


5. You have a DataFrame column for payment status from a Paytm-like app with mixed values: 'Yes', 'yes', 'Y', 'No', 'no', 'N', and some with extra spaces. Write code to unify this column so all paid statuses become 1 and all
unpaid statuses become 0, trimming whitespace and fixing capitalization where needed.

        Hint: Use str.strip(, str.lower0), and map/replace methods in pandas.

```python

data = {
    "payment_status": [
        "Yes", "yes", "Y", "No", "no", "N",
        " Yes ", " no ", "Y ", " N", "yes "
    ]
}

df = pd.DataFrame(data)

print("Before:")
print(df)

df["payment_status"] = df["payment_status"].str.strip().str.lower().map({
    "yes" :1,
    "y":1,
    "no" :0,
    "n":0
})

df
```
