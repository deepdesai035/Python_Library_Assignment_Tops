
4. Suppose you have a DataFrame of Instagram posts with a 'posted_at' column in UTC. Convert these timestamps to 'Asia/Kolkata' timezone and display the first 5 converted times.


```python

np.random.seed(42)

usernames = [
    "aman", "priya", "rahul", "neha", "jay",
    "simran", "rohit", "karan", "pooja", "vishal"
]

captions = [
    "Travel Diaries", "Gym Time", "Food Lover",
    "Weekend Vibes", "Nature Shot", "Coffee Break",
    "Work Mode", "Sunset Click", "Friends Forever",
    "New Adventure"
]

df = pd.DataFrame({
    "post_id": range(1001, 1101),
    "username": np.random.choice(usernames, 100),
    "caption": np.random.choice(captions, 100),
    "likes": np.random.randint(50, 5000, 100),
    "comments": np.random.randint(0, 500, 100),
    "posted_at": pd.to_datetime(
        np.random.choice(
            pd.date_range(
                start="2025-01-01",
                end="2025-03-31",
                freq="H",
                tz="UTC"
            ),
            100
        )
    )
})

print(df.head())
print(df.shape)
```

```python
df.info()
```

```python
df["time_zone_ist"] = df["posted_at"].dt.tz_convert("Asia/Kolkata")
```
