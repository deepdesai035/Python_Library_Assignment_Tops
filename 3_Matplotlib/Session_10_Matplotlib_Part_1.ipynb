```python
## Chatgpt

# Restaurant names
restaurants = [
    "Rajwadu", "Agashiye", "Barbeque Nation", "The Green House",
    "Tinello", "Gordhan Thal", "Tomato's", "Sankalp",
    "Jassi De Parathe", "Honest"
]

# Zomato ratings
ratings = [4.5, 4.7, 4.2, 4.4, 4.6, 4.3, 4.1, 4.0, 4.2, 3.9]

# Average meal price (₹)
prices = [800, 1200, 900, 700, 1500, 1000, 600, 500, 400, 300]

plt.figure(figsize=(8, 5))
plt.scatter(ratings, prices)

# Add restaurant names near points
for i in range(len(restaurants)):
    plt.annotate(restaurants[i], (ratings[i], prices[i]))

plt.xlabel("Zomato Rating")
plt.ylabel("Average Meal Price (₹)")
plt.title("Zomato Rating vs Average Meal Price of Restaurants")
plt.grid(True)

plt.show()
```

5. Customize a Matplotlib figure by creating a plot with two subplots: one line plot showing your daily Instagram screen time for a week, and one bar chart showing the number of posts you liked each day. Add
appropriate titles, axis labels, and save the figure as social_media_usage.png.

        Hint: Use plt.subplots() to create multiple axes in one figure.

```python
days = [1,2,3,4,5,6,7]
screen = [20,34,12,50,00,30,11]
likes = [100,33,21,33,99,22,0]

plt.subplot(1,2,1)
plt.plot(days , screen)
plt.xlabel("days")
plt.ylabel("screen time duration")
plt.title("Plot 1 - screen time ")

plt.subplot(1,2,2)
plt.bar(days , likes)
plt.xlabel("days")
plt.ylabel("likes")
plt.title("Plot 2 Likes ")

plt.savefig("social_media_usage.png")

plt.show
```
