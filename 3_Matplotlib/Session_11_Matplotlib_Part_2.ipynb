
1. Use plt.subplots() to create a 2x2 grid of subplots and plot four different types of charts (line, bar, scatter, and pie) using any sample data of your choice.


```python
number1 = np.random.randint(10,100,30)
number2 = np.random.randint(10,100,30)

fig,ax = plt.subplots(2,2)
ax[0,0].plot(number1, number2)
ax[0,1].bar(number1, number2)
ax[1,0].scatter(number1 , number2)
ax[1,1].pie(number1)

plt.show()
```


3. Create a multi-axis chart that shows the number of Instagram followers (left y-axis) and average daily posts (right y-axis) for five influencers. Use different colors and markers for each axis.


```python
fig, ax1 = plt.subplots(figsize=(8,5))

influ = ["deep", "Noor", "Hamza", "Love", "yami"]
follow = [1000, 200, 33, 11, 232]
posts = [2, 99, 8, 7, 0]

ax1.plot(influ, follow, marker='o', color='tab:blue')
ax1.set_ylabel('Followers', color='tab:blue')

ax2 = ax1.twinx()
ax2.plot(influ, posts, marker='s', color='tab:red')
ax2.set_ylabel('Posts per day', color='tab:red')

plt.show()
```


4. Plot the number of tickets sold for five recent Bollywood movies (categorical) versus their IMDB ratings (numeric) using a scatter plot. Add annotations to display the movie names above each point.


```python
movie = ["love" ,"dhurandhar", "HOTD","GOT","Breaking bad"]
ratings = [8,8.5,9,9.5,9.7]
tickets = [1000,1321,9894,9384,8390]

plt.scatter(ratings, tickets)

for i in range(len(movie)):
    plt.annotate(movie[i] , (ratings[i],tickets[i]))
    
```


5. Add a custom text annotation to a Matplotlib chart showing Flipkart's monthly sales, marking the highest sales point with the label 'Big Billion Days'.

        Hint: Use the ax.annotate() function to place the label at the correct data point.



```python
sales = np.random.randint(100,10000,12)
months = np.arange(1,13)

fig , ax2 = plt.subplots(figsize = (8,5))

ax2.scatter(months, sales)

max_sale = max(sales)

max_index = np.argmax(sales)

ax2.annotate(
    "Big Billion Days " ,
    xy = (months[max_index] , sales[max_index]),
    xytext =(months[max_index] , sales[max_index] +2000),
    arrowprops = dict(arrowstyle = "->")
)

plt.show()
```
