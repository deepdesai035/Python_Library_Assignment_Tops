
5. Use Seaborn's kdeplot to visualize the distribution of daily step counts for a week, as if analyzing data from a fitness app. Apply the 'whitegrid' theme and customize the plot color.

```python
step_count = np.random.randint(1000, 10000, 20)

#days = np.random.choice(20) in ["Mon" , "Tue", "Wed" , "Thurs" , "Fri" , "sat ", "sun"]
days = np.arange(1,21)

dataset = pd.DataFrame({
    "days" : days,
    "step_count" : step_count
})


def set_theme():
    sns.set(style = "whitegrid")
    sns.set_palette("deep")
    sns.set_context("notebook")

set_theme()

sns.kdeplot(dataset, x = days , y = step_count )
```
