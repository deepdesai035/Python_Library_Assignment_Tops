# SESSION 12 — Introduction to Seaborn

## What is Seaborn?

Seaborn is a **high-level data visualization library built on top of Matplotlib**.

---

## Why Seaborn is Used

Matplotlib provides flexibility but requires more code.

Seaborn provides:

• Better default aesthetics  
• Simplified syntax  
• Built-in statistical plots  
• Direct support for Pandas DataFrames  

---

## Key Advantages

• Attractive default themes  
• Easy handling of categorical data  
• Automatic statistical calculations  
• Integration with Pandas  

---

## When to Use Seaborn

• Exploratory Data Analysis (EDA)  
• Statistical visualization  
• Comparing distributions  
• Visualizing relationships  

---

## Conclusion

Seaborn simplifies complex visualizations and improves readability.

# Axes-Level vs Figure-Level Functions

## 1. Axes-Level Functions

Examples:

• histplot()  
• boxplot()  
• scatterplot()  

---

## Characteristics

• Work on a single plot  
• Return Axes object  
• Used for fine control  

---

## 2. Figure-Level Functions

Examples:

• relplot()  
• catplot()  
• displot()  
• pairplot()  

---

## Characteristics

• Can create multiple plots  
• Return Figure object  
• Higher-level abstraction  

---

## Key Difference

Axes-level → single plot control  
Figure-level → multiple plots and layouts

# Distribution Plots — histplot()

## Purpose

Visualize distribution of numerical data.

---

## Syntax

sns.histplot(data, x=None, bins=None, kde=False)

---

## Parameters

data → dataset  
x → column name  

bins (default=None)  
• Number of bars  

kde (default=False)  
• Adds smooth density curve  

---

## Use Case

• Understand spread of data  
• Identify skewness  

```python
sns.histplot(data=df, x="total_bill", bins=20, kde=True)
# sns.histplot(data=df, x="total_bill", bins=20)
plt.show()
```

# Stacked Bar Chart with More Than 2 Categories

## Concept

In stacked bar charts:

Each new layer is placed on top of the previous ones.

---

## Key Rule

For multiple categories:

bottom = cumulative sum of previous categories

---

## Example

If you have:

A, B, C

Then:

• A → base  
• B → bottom = A  
• C → bottom = A + B  

---

## Why This is Required

Because Matplotlib does not automatically stack multiple layers.

We must manually define stacking positions.

```python
import matplotlib.pyplot as plt
import numpy as np

months = ["Jan","Feb","Mar"]

product_A = np.array([100,150,200])
product_B = np.array([80,120,160])
product_C = np.array([50,100,120])

# First layer
plt.bar(months, product_A, label="A")

# Second layer (stacked on A)
plt.bar(months, product_B, bottom=product_A, label="B")

# Third layer (stacked on A + B)
plt.bar(months, product_C, bottom=product_A + product_B, label="C")

plt.title("Stacked Bar Chart (3 Categories)")
plt.legend()
plt.show()
```
