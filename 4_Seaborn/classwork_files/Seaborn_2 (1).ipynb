# SESSION 9 — Regression Plots in Seaborn

## What is a Regression Plot?

A regression plot shows:

• Relationship between two variables  
• Best-fit line (trend line)  

---

## Why Use Regression Plots?

• Understand correlation  
• Predict trends  
• Identify patterns  

---

## Two Main Functions

• sns.regplot() → Axes-level  
• sns.lmplot() → Figure-level  

# sns.regplot()

## Syntax

sns.regplot(x=None, y=None, data=None)

---

## Parameters

x → independent variable  
y → dependent variable  
data → dataset  

---

## Default Behavior

• Adds regression line automatically  
• Shows scatter + line  

---

## Use Case

• Quick relationship visualization  

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

sns.regplot(x="total_bill", y="tip", data=df)

plt.title("Regplot Example")
plt.show()
```

# sns.lmplot()

## Syntax

sns.lmplot(x=None, y=None, data=None, col=None)

---

## Parameters

x → independent variable  
y → dependent variable  
data → dataset  

col → creates multiple plots (faceting)  

---

## Difference from regplot

regplot → single plot  
lmplot → multiple plots (faceted)  

---

## Use Case

• Compare trends across categories  

```python
sns.lmplot(x="total_bill", y="tip", data=df, col="sex")

plt.show()
```

# Lowess Smoothing

## What is Lowess?

A smooth curve instead of straight line.

---

## Syntax

sns.regplot(..., lowess=True)

---

## Use Case

• Non-linear relationships  

```python
sns.regplot(x="total_bill", y="tip", data=df, lowess=True)

plt.title("Lowess Smoothing")
plt.show()
```

# SESSION 10 — Advanced Seaborn Styling

## Why Styling is Important

Default charts are basic.

Styling improves:

• Readability  
• Presentation  
• Professional quality  

---

## Key Tools

• sns.set()  
• Color palettes  
• Context settings  

# sns.set()

## Syntax

sns.set(style=None)

---

## Styles

• darkgrid  
• whitegrid  
• white  
• ticks  

---

## Use Case

Set global style for all plots

```python
sns.set(style="white")

sns.histplot(data=df, x="total_bill")

plt.show()
```

# Color Palettes

## Syntax

sns.set_palette("palette_name")

---

## Common Palettes

• deep  
• muted  
• colorblind  

---

## Use Case

• Improve visual clarity  
• Accessibility  

```python
sns.set_palette("colorblind")

sns.countplot(x="day", data=df)

plt.show()
```

# Context Settings

## Syntax

sns.set_context("context")

---

## Options

• notebook (default)  
• talk  
• poster  

---

## Purpose

Control scaling of:

• Text  
• Lines  

```python
sns.set_context("notebook")

sns.regplot(x="total_bill", y="tip", data=df)

plt.show()
```

# Corporate Style Demo

Combine:

• Style  
• Palette  
• Context  

```python
sns.set(style="whitegrid")
sns.set_palette("muted")
sns.set_context("talk")

sns.barplot(x="day", y="total_bill", data=df)

plt.title("Corporate Styled Chart")
plt.show()
```

# SESSION 11 — Exporting High-Quality Charts

# Template Styling

## Concept

Create reusable styling setup

---

## Example

Set style + palette once  
Reuse across charts  

```python
def set_theme():
    sns.set(style="whitegrid")
    sns.set_palette("deep")
    sns.set_context("notebook")

set_theme()

sns.lineplot(x="total_bill", y="tip", data=df)

plt.show()
```
